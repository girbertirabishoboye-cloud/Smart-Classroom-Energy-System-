```python
import pandas as pd

# Load the files without treating the first row as column names
light_df = pd.read_csv(
    "/kaggle/input/datasets/ranakrc/smart-building-system/KETI/413/light.csv",
    header=None,
    names=["timestamp", "light_level"]
)

pir_df = pd.read_csv(
    "/kaggle/input/datasets/ranakrc/smart-building-system/KETI/413/pir.csv",
    header=None,
    names=["timestamp", "motion"]
)

print("LIGHT SENSOR")
print(light_df["light_level"].describe())
print("Unique light values:", light_df["light_level"].nunique())

print("\nPIR MOTION SENSOR")
print(pir_df["motion"].describe())
print("Unique motion values:", sorted(pir_df["motion"].unique()))

print("\nMotion value counts:")
print(pir_df["motion"].value_counts())

print("\nFirst light records:")
display(light_df.head())

print("\nFirst PIR records:")
display(pir_df.head())
```

```python
import pandas as pd

# Convert timestamps to readable dates
light_df["datetime"] = pd.to_datetime(
    light_df["timestamp"], unit="s", utc=True
)

pir_df["datetime"] = pd.to_datetime(
    pir_df["timestamp"], unit="s", utc=True
)

print("LIGHT SENSOR RECORDING PERIOD")
print("Start:", light_df["datetime"].min())
print("End:", light_df["datetime"].max())

print("\nPIR MOTION SENSOR RECORDING PERIOD")
print("Start:", pir_df["datetime"].min())
print("End:", pir_df["datetime"].max())

print("\nLIGHT SENSOR DUPLICATE TIMESTAMPS:")
print(light_df["timestamp"].duplicated().sum())

print("\nPIR SENSOR DUPLICATE TIMESTAMPS:")
print(pir_df["timestamp"].duplicated().sum())
```

```python
import pandas as pd

# Sort both datasets by timestamp
light_df = light_df.sort_values("timestamp").copy()
pir_df = pir_df.sort_values("timestamp").copy()

# Keep the required sensor columns
light_data = light_df[["timestamp", "light_level"]].copy()

pir_data = pir_df[["timestamp", "motion"]].copy()
pir_data = pir_data.rename(
    columns={"timestamp": "pir_timestamp"}
)

# Align the nearest PIR reading to each light reading
merged_df = pd.merge_asof(
    light_data,
    pir_data,
    left_on="timestamp",
    right_on="pir_timestamp",
    direction="nearest",
    tolerance=10
)

# Remove rows without a matched motion reading
merged_df = merged_df.dropna(subset=["motion"]).copy()

# Convert motion into binary values
merged_df["motion_detected"] = (
    merged_df["motion"] > 0
).astype(int)

# Convert timestamps to readable datetime
merged_df["datetime"] = pd.to_datetime(
    merged_df["timestamp"],
    unit="s",
    utc=True
)

print("MERGED DATASET")
print("Shape:", merged_df.shape)

print("\nFirst 10 records:")
display(merged_df.head(10))

print("\nMissing values:")
print(merged_df.isnull().sum())

print("\nMotion detection distribution:")
print(merged_df["motion_detected"].value_counts())

print("\nLight level statistics:")
print(merged_df["light_level"].describe())
```

```python
import pandas as pd

# Convert timestamps to South Korea local time (KETI dataset location)
merged_df["local_datetime"] = (
    merged_df["datetime"].dt.tz_convert("Asia/Seoul")
)

# Extract hour
merged_df["hour"] = merged_df["local_datetime"].dt.hour

# Create recommendation target
merged_df["light_status"] = (
    (merged_df["motion_detected"] == 1) &
    (merged_df["hour"].between(7, 18)) &
    (merged_df["light_level"] < 100)
).astype(int)

# Convert numeric labels into readable labels
merged_df["light_status"] = merged_df["light_status"].map({
    1: "ON",
    0: "OFF"
})

print("TARGET DISTRIBUTION")
print(merged_df["light_status"].value_counts())

print("\nTarget percentages:")
print(
    merged_df["light_status"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nSample dataset:")
display(
    merged_df[
        ["light_level", "motion_detected", "hour", "light_status"]
    ].head(10)
)
```

```python
from sklearn.model_selection import train_test_split

# Select model input features
X = merged_df[
    ["light_level", "motion_detected", "hour"]
]

# Select target
y = merged_df["light_status"]

# Split dataset: 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("DATA SPLIT RESULTS")
print("Total records:", len(merged_df))
print("Training records:", len(X_train))
print("Testing records:", len(X_test))

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())

print("\nFeature columns:")
print(X.columns.tolist())
```

```python
from sklearn.metrics import precision_recall_curve
from sklearn.tree import DecisionTreeClassifier
import numpy as np

# Divide training data chronologically:
# 80% for fitting and 20% for validation
validation_split = int(len(X_train) * 0.80)

X_fit = X_train.iloc[:validation_split]
y_fit = y_train.iloc[:validation_split]

X_val = X_train.iloc[validation_split:]
y_val = y_train.iloc[validation_split:]

# Train a temporary Decision Tree
tuning_model = DecisionTreeClassifier(
    max_depth=5,
    random_state=42,
    class_weight="balanced"
)

tuning_model.fit(X_fit, y_fit)

# Get motion probabilities on validation data
motion_index = list(tuning_model.classes_).index(1)

val_probabilities = tuning_model.predict_proba(X_val)[:, motion_index]

# Calculate precision and recall at different thresholds
precision, recall, thresholds = precision_recall_curve(
    y_val,
    val_probabilities
)

# Calculate F1-score for each threshold
f1_scores = (
    2 * precision[:-1] * recall[:-1] /
    (precision[:-1] + recall[:-1] + 1e-10)
)

# Select threshold with the highest validation F1-score
best_index = np.argmax(f1_scores)
best_threshold = thresholds[best_index]

print("THRESHOLD TUNING RESULTS")

print("\nValidation observations:", len(X_val))
print("Best probability threshold:", round(best_threshold, 4))
print("Validation precision:", round(precision[best_index], 4))
print("Validation recall:", round(recall[best_index], 4))
print("Validation F1-score:", round(f1_scores[best_index], 4))
```

```python
import pandas as pd
import numpy as np

# Inspect predicted motion probabilities
probability_summary = pd.Series(val_probabilities).describe(
    percentiles=[0.25, 0.50, 0.75, 0.90, 0.95, 0.99]
)

print("VALIDATION MOTION PROBABILITY DISTRIBUTION")
display(probability_summary)

print("\nUnique probability values:")
print(np.sort(np.unique(val_probabilities)))

print("\nActual validation class distribution:")
print(y_val.value_counts())
```

```python

from sklearn.metrics import precision_score, recall_score, f1_score

thresholds_to_test = [
    0.005,
    0.10,
    0.15,
    0.20,
    0.23,
    0.50,
    0.90
]

print("THRESHOLD COMPARISON ON VALIDATION DATA")

for threshold in thresholds_to_test:

    val_predictions = (
        val_probabilities >= threshold
    ).astype(int)

    precision_value = precision_score(
        y_val, val_predictions, zero_division=0
    )

    recall_value = recall_score(
        y_val, val_predictions, zero_division=0
    )

    f1_value = f1_score(
        y_val, val_predictions, zero_division=0
    )

    print(
        f"Threshold: {threshold:.3f} | "
        f"Precision: {precision_value:.3f} | "
        f"Recall: {recall_value:.3f} | "
        f"F1-score: {f1_value:.3f}"
    )
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import precision_recall_curve
import numpy as np

# Split training data chronologically: 80% fitting, 20% validation
split_index = int(len(X_train) * 0.80)

X_fit = X_train.iloc[:split_index]
X_val = X_train.iloc[split_index:]

y_fit = y_train.iloc[:split_index]
y_val = y_train.iloc[split_index:]

# Train a separate Decision Tree
tuning_model = DecisionTreeClassifier(
    max_depth=5,
    random_state=42,
    class_weight="balanced"
)

tuning_model.fit(X_fit, y_fit)

# Get motion probabilities on validation data
val_probabilities = tuning_model.predict_proba(X_val)[:, 1]

# Calculate precision and recall at different thresholds
precision, recall, thresholds = precision_recall_curve(
    y_val,
    val_probabilities
)

# Calculate F1-score for each threshold
f1_scores = (
    2 * precision[:-1] * recall[:-1]
    / (precision[:-1] + recall[:-1] + 1e-10)
)

# Find threshold with the highest validation F1-score
best_index = np.argmax(f1_scores)
best_threshold = thresholds[best_index]

print("VALIDATION THRESHOLD ANALYSIS")
print("--------------------------------")

print("Fitting data:", X_fit.shape)
print("Validation data:", X_val.shape)

print("\nBest threshold:", round(best_threshold, 4))
print("Validation precision:", round(precision[best_index], 4))
print("Validation recall:", round(recall[best_index], 4))
print("Validation F1-score:", round(f1_scores[best_index], 4))

print("\nActual validation distribution:")
print(y_val.value_counts())

print("\nUnique predicted probabilities:")
print(np.unique(val_probabilities))
```

```python

# ============================================================
# SMART CLASSROOM ENERGY SYSTEM
# COMPLETE MODEL COMPARISON, VALIDATION AND FINAL TESTING
# Corrected KETI Room 413 dataset
# ============================================================

import numpy as np
import pandas as pd
import joblib
import warnings

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    precision_recall_curve
)

warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# 1. PREPARE DATA
# ------------------------------------------------------------

features = [
    "light_level",
    "temperature",
    "humidity",
    "co2",
    "hour"
]

X = corrected_data[features].copy()
y = corrected_data["motion_detected"].copy()

# Chronological 80% training / 20% testing
split_index = int(len(corrected_data) * 0.80)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

# Split training into fitting and validation periods
fit_index = int(len(X_train) * 0.80)

X_fit = X_train.iloc[:fit_index]
X_val = X_train.iloc[fit_index:]

y_fit = y_train.iloc[:fit_index]
y_val = y_train.iloc[fit_index:]

print("=" * 60)
print("DATA PREPARATION")
print("=" * 60)

print("Fitting:", X_fit.shape)
print("Validation:", X_val.shape)
print("Testing:", X_test.shape)

print("\nFitting period:")
print(corrected_data["datetime"].iloc[0])
print("to", corrected_data["datetime"].iloc[fit_index - 1])

print("\nValidation period:")
print(corrected_data["datetime"].iloc[fit_index])
print("to", corrected_data["datetime"].iloc[split_index - 1])

print("\nTesting period:")
print(corrected_data["datetime"].iloc[split_index])
print("to", corrected_data["datetime"].iloc[-1])

print("\nMotion percentage:")
print("Fitting:", round(y_fit.mean() * 100, 2), "%")
print("Validation:", round(y_val.mean() * 100, 2), "%")
print("Testing:", round(y_test.mean() * 100, 2), "%")


# ------------------------------------------------------------
# 2. DEFINE MODELS
# ------------------------------------------------------------

models = {
    "Decision Tree Depth 3": DecisionTreeClassifier(
        max_depth=3,
        random_state=42,
        class_weight="balanced"
    ),

    "Decision Tree Depth 5": DecisionTreeClassifier(
        max_depth=5,
        random_state=42,
        class_weight="balanced"
    ),

    "Decision Tree Depth 8": DecisionTreeClassifier(
        max_depth=8,
        random_state=42,
        class_weight="balanced"
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        max_depth=10,
        min_samples_leaf=5,
        class_weight="balanced_subsample",
        random_state=42,
        n_jobs=-1
    ),

    "Extra Trees": ExtraTreesClassifier(
        n_estimators=200,
        max_depth=10,
        min_samples_leaf=5,
        class_weight="balanced_subsample",
        random_state=42,
        n_jobs=-1
    ),

    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", LogisticRegression(
            class_weight="balanced",
            max_iter=1000,
            random_state=42
        ))
    ])
}


# ------------------------------------------------------------
# 3. TRAIN AND COMPARE MODELS ON VALIDATION DATA
# ------------------------------------------------------------

results = []
trained_models = {}
model_thresholds = {}

for name, model in models.items():

    print("\nTraining:", name)

    model.fit(X_fit, y_fit)

    probabilities = model.predict_proba(X_val)[:, 1]

    precision, recall, thresholds = precision_recall_curve(
        y_val,
        probabilities
    )

    f1_values = (
        2 * precision[:-1] * recall[:-1]
        / (precision[:-1] + recall[:-1] + 1e-10)
    )

    best_index = np.argmax(f1_values)
    best_threshold = thresholds[best_index]

    predictions = (
        probabilities >= best_threshold
    ).astype(int)

    results.append({
        "Model": name,
        "Threshold": best_threshold,
        "Accuracy": accuracy_score(y_val, predictions),
        "Precision": precision_score(
            y_val, predictions, zero_division=0
        ),
        "Recall": recall_score(
            y_val, predictions, zero_division=0
        ),
        "F1 Score": f1_score(
            y_val, predictions, zero_division=0
        ),
        "Predicted Motion": int(predictions.sum())
    })

    trained_models[name] = model
    model_thresholds[name] = best_threshold


# ------------------------------------------------------------
# 4. DISPLAY MODEL COMPARISON
# ------------------------------------------------------------

comparison = pd.DataFrame(results)

comparison = comparison.sort_values(
    by="F1 Score",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 60)
print("VALIDATION MODEL COMPARISON")
print("=" * 60)

print(comparison.to_string(index=False))

# Select model with the highest validation F1-score
best_model_name = comparison.iloc[0]["Model"]

best_model = trained_models[best_model_name]
best_threshold = model_thresholds[best_model_name]

print("\nSELECTED MODEL:", best_model_name)
print("Selected threshold:", round(best_threshold, 6))


# ------------------------------------------------------------
# 5. FINAL EVALUATION ON UNTOUCHED TEST DATA
# ------------------------------------------------------------

# Keep the selected model trained on fitting data.
# This ensures its validation-selected threshold is used
# with the same fitted model.

test_probabilities = best_model.predict_proba(X_test)[:, 1]

test_predictions = (
    test_probabilities >= best_threshold
).astype(int)

test_accuracy = accuracy_score(y_test, test_predictions)

test_precision = precision_score(
    y_test, test_predictions, zero_division=0
)

test_recall = recall_score(
    y_test, test_predictions, zero_division=0
)

test_f1 = f1_score(
    y_test, test_predictions, zero_division=0
)

cm = confusion_matrix(
    y_test,
    test_predictions,
    labels=[0, 1]
)

print("\n" + "=" * 60)
print("FINAL TEST RESULTS")
print("=" * 60)

print("Selected model:", best_model_name)
print("Decision threshold:", round(best_threshold, 6))

print("\nAccuracy:", round(test_accuracy, 4))
print("Precision:", round(test_precision, 4))
print("Recall:", round(test_recall, 4))
print("F1-score:", round(test_f1, 4))

print("\nConfusion Matrix:")
print(cm)

print("\nClassification Report:")
print(classification_report(
    y_test,
    test_predictions,
    labels=[0, 1],
    target_names=["No Motion", "Motion Detected"],
    zero_division=0
))


# ------------------------------------------------------------
# 6. FEATURE IMPORTANCE
# ------------------------------------------------------------

if hasattr(best_model, "feature_importances_"):

    importance_values = best_model.feature_importances_

elif isinstance(best_model, Pipeline):

    importance_values = np.abs(
        best_model.named_steps["classifier"].coef_[0]
    )

else:

    importance_values = None


if importance_values is not None:

    importance_df = pd.DataFrame({
        "Feature": features,
        "Importance": importance_values
    }).sort_values(
        by="Importance",
        ascending=False
    )

    print("\n" + "=" * 60)
    print("FEATURE IMPORTANCE")
    print("=" * 60)

    print(importance_df.to_string(index=False))


# ------------------------------------------------------------
# 7. SAVE SELECTED MODEL
# ------------------------------------------------------------

final_package = {
    "model": best_model,
    "features": features,
    "threshold": float(best_threshold),
    "model_name": best_model_name,
    "target": "motion_detected"
}

model_path = "/kaggle/working/smart_classroom_final_model.joblib"

joblib.dump(final_package, model_path)

print("\n" + "=" * 60)
print("MODEL SAVED SUCCESSFULLY")
print("=" * 60)

print("File:", model_path)

print("\nIMPORTANT:")
print("This model predicts PIR motion activity.")
print("It does not directly predict actual light ON/OFF status.")
```
